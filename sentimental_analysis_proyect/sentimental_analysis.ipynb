### Primera Funcion

```python
import csv
import pandas
```

```python
#Nueva 1
def is_valid_word(word:str) -> bool:
    punctuation_chars = ["'", '"', ",", ".", "!", ":", ";", "#", "@"]
    return not any(char in word for char in punctuation_chars)

#Nueva 2
def clean_word(word:str) -> str | None:
    punctuation_chars = ["'", '"', ",", ".", "!", ":", ";", "#", "@"] 
    clean_word = word
    for punctuation in punctuation_chars:
        clean_word = clean_word.replace(punctuation, "").strip()      
    return clean_word

#Nueva 3
def load_words_from_file(filepath: str) -> list[str]:
    words = []
    with open(filepath, "r", newline="") as file:
        for line in file:
            if line[0] != ';' and line[0] != '\n' and line[0] != "":
                words.append(str(line).strip())
    return words

#Nueva 4
def get_pos_word(raw_word: str) -> int:
    return sum(1 for raw in raw_word.lower().split() if raw == clean_word(raw_word))

#Nueva 5
def get_neg_word(raw_word: str) -> int:
    return sum(1 for raw in raw_word.lower().split() if raw == clean_word(raw_word))

#Nueva 6
def analyze_tweets(input_filepath: str, output_filepath: str) -> None:
    fieldnames = ["Tweet, Number of Retweets","Number of Replies", "Positive Score", "Negative Score", "Net Score"]
    with open(input_filepath, newline="") as in_file, \
        open(output_filepath, "w", newline="") as out_file:
        csv_reader = csv.DictReader(in_file)
        csv_writer = csv.DictWriter(out_file, fieldnames=fieldnames)
        csv_writer.writeheader()

        for row in csv_reader:
            if is_valid_word(row.get("tweet_text","").split()):
                pass
            
```
